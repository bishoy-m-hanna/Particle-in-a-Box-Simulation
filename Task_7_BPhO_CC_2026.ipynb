### **Introduction**
In 1924, **Louis de Broglie** proposed a radical symmetry in nature: if light waves could exhibit particle-like behavior (photons), then material particles (like electrons) must exhibit wave-like properties. He quantified this using the **de Broglie relation**, which connects which connects a particle's momentum ($p$) to its wavelength ($\lambda$) via Planck’s constant ($h$):

$$\lambda = \frac{h}{p}$$

While de Broglie established the existence of matter waves and proposed a relation to find their wavelengths, he didn't try to find an equation that governs their behavior, but someone else did in 1925.

In late 1925, **Erwin Schrödinger**, an Austrian physicist, wanted to find a wave equation for the hydrogen atom. He treated the electron as a physical wave oscillating in space and time. After heavy mathematical operations, involving calculus and waves, and scientific assumptions, Schrödinger came with his final result.

Interestingly, Schrödinger wasn't the first to solve the quantum puzzle. Just months earlier, **Werner Heisenberg** created the **Matrix Mechanics**. However, this version used complex matrices. It was abstract and difficult to visualize, that it was hated by physicists at the time. Eventually, Schrödinger proved that his wave math and Heisenberg’s matrix math were actually mathematically equivalent, but different languages describing the same result.

It is important to note that the **Time-Dependent Schrödinger Equation (TDSE)** cannot be strictly proved or derived from first principles and deeper laws in the same way one derives a geometric theorem, since it is a postulate, which means a fundamental assertion (assumption) declared to be true. Physicists can't prove a postulate using logic alone, but they must test its predictions against nature until it is either validated or refuted by **experimental reality & results**. Therefore, the Schrödinger equation persisted to be a hypothesis until experiments validated its predictions, proving they aligned perfectly with physical results. Once confirmed, it was universally accepted by scientists. Thus, we treat the Schrödinger equation as a **foundational starting point** from which all other quantum mechanical behaviors are derived.

### **Derivation of TDSE**
While Schrödinger's oringinal path involved heavy, non-intuitive mathematics, relying on the formalisms of Hamiltonian mechanics, we can arrive at the same equation through a heuristic approach that reveals the underlying logic.

Because the electron behaves like a wave, to describe this wave, we need a **wave function**. Since any valid wave in the universe must satisfy the fundamental rules of wave mechanics, our wave function **must** be a solution to the **Classical Wave Equation**:

$$\frac{\partial^2 \, y(x,t)}{\partial\, x^2} = \frac{1}{v^2}\frac{\partial^2 \, y(x,t)}{\partial t^2}$$

Yet, because there are many valid solutions that describe a wave propagating to the right we choose the follwing familiar standard harmonic wave function as our starting point to keep things simple:

$$y(x,t) = A\cos(kx - \omega t)$$

According to de Broglie-Einstein postulates:

$$\lambda = \frac{h}{p} \quad\quad E = hf$$

Constant $p$ means constant $\lambda$, and therefore constant $f$ and $E$. However, when a force acts on the electron, its momentum changes according to Newton's second law:

$$F = \frac{dp}{dt}$$

This means that wavelength is not constant and changes with respect to potential energy ($V$). Hence, we need a function more complex than ($y$) to describe the wave. Therefore, we need to develop an equation that tells us the wave function if we tell it information about the forces acting on a particle. As we have seen with the classical wave equation, the most common form that has a function as a solution is a differential equation. The equation must:
1. be consistent with the de Broglie-Einstein postulates.
2. be consistent with the energy equation for a particle:

$$E = K + V \quad\quad K = \frac{1}{2}mv^2 = \frac{p^2}{2m}$$

$$E = \frac{p^2}{2m} + V$$

3. be linear in $\psi(x,t)$

    If $\psi_1(x,t)$ and $\psi_2(x,t)$ are two different solutions, then arbitrary linear combinations, such as $\psi(x,t) = c_1\psi_1(x,t) + c_2\psi_2(x,t)$, are also solutions.
  
4. For a **constant potential**, the desired differential equation will have sinusoidal traveling wave solutions of **constant wavelength and frequency**.

    Because, if the potential energy is constant, then there are no forces acting on the wave, since

    $$F = -\frac{\partial \,V(x,t)}{\partial \, x}$$

    which also means that momentum is constant according to Newton's second law discussed above. Therefore, $\lambda$ and $f$ are constant.

As we have said, the wave equation we are trying to derive must be consistent with de Broglie-Einstein posulates and the energy of a particle (conservation of energy). Let's try combining those equations:

$$E = \frac{p^2}{2m} + V \quad\quad p = \frac{h}{\lambda} \quad\quad E = hf$$

$$hf = \frac{h^2}{2m\lambda^2} + V(x,t)$$

We add $\omega = 2\pi f$ and $k = \frac{2\pi}{\lambda}$ to the equation, since they are used in the wave function, and we will learn the connection soon. Also, as we do this, we add a lot of $2\pi$, which will become annoying as we progress; therefore, we hide them with the reduced Planck's constant $\hbar = \frac{h}{2\pi}$:

$$\quad\quad \hbar \, \boxed{\omega} = \frac{\hbar^2 \,\boxed{k^2}}{2m} + V(x,t) \quad\quad \text{[Eq. 3]}$$

If we take a look at the function $y(x,t) = A\cos(kx - wt)$, we observe that if we take the second derivative with respect to $x$, we get $-k^2$; if we take the first derivative with respect to $t$, we get $\omega$. This is the first hint that we need a sinusoidal wave function.

As the fourth assumption suggests, when there potential energy is constant, the differential equation has sinusoidal wave solutions. This is the second hint.

These hints suggest that $\psi$ is valid:

$$\psi(x,t) = A\cos(kx-\omega t)$$

Combining $\psi$ with the partial derivatives $\frac{\partial^2 \,\psi(x,t)}{\partial\, x^2}$ and $\frac{\partial \,\psi(x,t)}{\partial\, t}$, adding the potential energy function, and multiplying it by $\psi$ to ensure linearity (since differentiating sinusoidal functions gives sinusoidal functions in the result), results in final expression of the wave's hamiltonian:

$$\beta \frac{\partial \,\psi(x,t)}{\partial\, t} = \alpha \frac{\partial^2 \,\psi(x,t)}{\partial\, x^2} + V(x,t)\psi(x,t)$$

When assuming $V(x,t) = V_0$ (constant):

$$\beta \omega \sin(kx-\omega t) = -\alpha k^2\cos(kx - \omega t) + V_0\cos(kx - \omega t)$$

$$\therefore \beta \omega \sin(kx-\omega t) + (\alpha k^2 - V_0)\cos(kx - \omega t) = 0$$

From the final expression, to always get a result of zero, both $\beta$ and $\alpha k^2 - V_0$ must be zero, because $\sin$ and $\cos$ are independent. The first reason this is impossible is becasue this would eliminate the time derivative:

$$0 = -\alpha k^2\cos(kx - \omega t) + V_0\cos(kx - \omega t)$$

This eqation has no time dependence, which means that the wave is frozen.It cannot propagate, cannot have frequency, and cannot describe a moving electron.

The second reason is that $k^2 = \frac{V_0}{\alpha}$, which bonds the wavelength, thus momentum, only to the potential energy

$$p = \hbar \sqrt{\frac{V_0}{\alpha}}$$

Nevertheless, wavelength depends on momentum, which depends on kinetic energy:

$$p = \sqrt{2m(E-V)} = \sqrt{2mK}$$

$$\sqrt{2mK} \ne \hbar \sqrt{\frac{V_0}{\alpha}}$$

This is wrong. The original problem is that there is a $\sin$ and a $\cos$ in the same equation, which prevents substituting with any value inside thems to get a zero. The mix results from differentiating once with respect to time and twice with respect to space.

If a single sinusoidal function doesn't work, and we know that wave functions are linear, then why don't we add two functions together, one with a $\sin$, and the other with a $\cos$? Like this, when we differentiate, each one will become the other and the only problem that remains is handling the signs.

$$\psi(x,t) = A\cos(kx - \omega t) + B\sin(kx - \omega t)$$

$$\frac{\partial}{\partial t} \cos(kx - \omega t) = \omega\sin(kx - \omega t)$$

$$\frac{\partial}{\partial t} \sin(kx - \omega t) = -\omega\cos(kx - \omega t)$$

$$\frac{\partial^2}{\partial x^2} \cos(kx - \omega t) = -k^2\cos(kx - \omega t)$$

$$\frac{\partial^2}{\partial x^2} \sin(kx - \omega t) = -k^2\sin(kx - \omega t)$$

Therefore,

$$\beta \frac{\partial\psi}{\partial t} = \beta\omega A\sin(kx - \omega t) - \beta\omega B\cos(kx - \omega t)$$

$$\alpha\frac{\partial^2\psi}{\partial x^2} = -\alpha k^2 A\cos(kx-\omega t) - \alpha k^2 B \sin(kx - \omega t)$$

$$V_0\psi = V_0 A \cos(kx - \omega t) + V_0 B \sin(kx - \omega t)$$

Since the functions are linear, we deal with the terms with sines independently from those of cosines. Also, becasue all sines are equal, we factor them out. We do the same with cosines. This results in two equations:

$$\beta\omega A = -\alpha k^2 B + V_0B ⇒ \beta\omega A = (V_0 - \alpha k^2)B$$

$$-\beta\omega B = -\alpha k^2 A + V_0A ⇒ -\beta\omega B = (V_0 - \alpha k^2)A$$

Because $A$ is dependent on $B$ and vice versa, if we try to solve them using **regular numbers**, we are forced back to zero. What about **imaginary numbers**? Let's substitute $B = iA$ into both equations:

$$\beta\omega A = (V_0 - \alpha)(iA)$$

$$\boxed{\beta\omega = i(V_0 - \alpha k^2)}$$

$$-\beta\omega (iA) = (V_0 - \alpha k^2)A$$

$$-i\beta\omega = V_0 - \alpha k^2$$

We multiply by $i$:

$$-i^2\beta\omega = i(V_0 - \alpha k^2)$$

Since $i^2 = -1$

$$\boxed{\beta\omega = i(V_0 - \alpha k^2)}$$

Equations are identical when $\boxed{B = iA}$:

$$\psi(x,t) = A\cos(kx - \omega t) + iA\sin(kx - \omega t)$$

$$\psi(x,t) = A[\cos(kx - \omega t) + i\sin(kx - \omega t)]$$

According to Euler:

$$e^{i\theta} = \cos\theta+i\sin\theta$$

When we apply this on $\psi$:

$$\psi(x,t) = Ae^{i(kx-wt)}$$

**NOTE:** If you want to set $A = iB$, it would work. Yet, you would factor out $iB$ if you want to put it in Euler's form, which is easier to handle.

The final form of the wave equation is:

$$\frac{\beta\omega}{i} Ae^{i(kx - \omega t)} = -\alpha k^2Ae^{i(kx - \omega t)} + V_0Ae^{i(kx - \omega t)}$$

We factor out $Ae^{i(kx - \omega t)}$:

$$\frac{\beta\omega}{i} = -\alpha k^2 + V_0$$

When we compare this result with **Eq. 3**, we find:

$$-\alpha k^2 = \frac{\hbar^2k^2}{2m} ⇒ \alpha = -\frac{\hbar^2}{2m}$$

$$\frac{\beta\omega}{i} = \hbar\omega \Rightarrow \beta = i\hbar$$

Finally, the Tme-Dependent Schrodinger Equation is:

$$\color{#89CFF0}{\bbox[10px, border: 3px solid #89CFF0]{i\hbar\frac{\partial \psi(x,t)}{\partial t} = -\frac{\hbar^2}{2m}\frac{\partial^2\psi(x,t)}{\partial x^2}+ V(x,t)\psi(x,t)}}$$



























































### **The Wave Function**
The solution to Schrodinger's Equation is a complex wave function. Let's learn more complex numbers.

A complex number $z$ can be expressed as:

$$z = a + ib$$

where $a$ is the real part and $b$ is the imaginary part:

$$Re(z) = a \quad\quad Im(z) = b$$

$a$ and $b$ are separated; we can treat them as coordinates. Therefore, we can create a plane where the real numbers are on the horizontal axis and the imaginary numbers are on the vertical axis. Then, $z$ can be represented as the point $(a, b)$.

If we start from the center of this plane, move horizontally to $a$, move vertically to $b$, then move diagonally to the center again, we create a right-angled triangle with a diagonal $r$, connecting $(0, 0)$ and $(a, b)$.

To find $r$, we use an operation called the **modulus** of a complex number, denoted by $|z|$ and utilizes the Pythagorean theorem:

$$|z| = \sqrt{a^2 + b^2}$$

We can also use the **conjugate** to find $r$. The conjugate is a complex number its imaginary part sign is reversed, denoted to as $\bar{z}$.

$$z = a + bi \quad \quad \bar{z} = a - bi$$

To find $r^2$, we multiply $z\cdot\bar{z}$:

$$|z|^2 = z\cdot \bar{z} \qquad\qquad$$

$$\qquad = (a + bi)(a - bi)$$

$$ = a^2- i^2b^2 \quad$$

$$ = a^2 + b^2 \qquad$$

Same result as above. The most important note about the **modulus square** or $|z|^2$ is that it is a **real number**.

What is the similarity between $z$ and $\psi$?

From Pythagorean theorem and trig functions, when:

$$z = r\left[\cos(\theta) + i\sin(\theta)\right]\qquad r = A \qquad \theta = kx - \omega t$$

$$\therefore \psi(x,t) = A[\cos(kx - \omega t) + i\sin(kx - \omega t)]$$

It is woth mentioning that complex quantities cannot be measured by physics instrument. Therefore, $\psi$ is not measurable. However, the modulus is indeed a real number, which means that it is measurable.

In 1926, Max Born suggested that the modulus square of the wave function $|\psi|^2$ is the **probability density**, which is the probability per unit length in 1D or per unit volume in 3D of finding an electron at time $t$ between $x$ and $x + dx$.The reason why you should look at an interval in space not a point is that the probability at a single mathematical point is zero. Therefore, to find the probability within an infinitesimally small interval $dx$, we multiply $dx$, which could be considered the width, by $|\psi(x,t)|^2$, which could be considered the length of a very thin rectangly undert he curve of the function $|\psi(x,t)|^2$. Yet, if you want to find the probability density over a larger inverval from $a$ to $b$, you should take the integral of the modulus with respect to $dx$:

$$\color{#89CFF0}{\bbox[5px, border: 3px solid]{P(x,t) = \int_a^b |\psi(x,t)|^2dx}}$$

And of course

$$a \le x \le b$$

There is an important distinction between the **probability** and the **probability density**. The probability function $f(b)$ returns the probability of finding an electron between $a$ and $b$, where $a$ is the beginning of the space interval. It is the accumulation of probabilities from $a$ to $b$. To better understand this, imagine an iron rod, for which you made a function that if you give it the length of a piece of this rod, it returns the total mass. The function starts from point zero, the beginning of the rod, and moves to $b$. This is the accumulated mass, which is analogous to the probability function. If you want the probability of an interval that doesn't start from $a$ to $b$, but from $c$ to $b$, where $c$ is somewhere in the middle, you find $f(b)$ from $a$ to $b$, then subtract from it $f(c)$ from $a$ to $c$.

On the other hand, the probability density function is the derivative of the probability function $f(x)$ with respect to length in 1D space (volume in 3D space) where the electron may exist. This is similar to the linear density function that is the derivative of mass with respect to length. As we know from calculus, if we have the derivative function (probability density) and we want the probability we integrate from $a$ to $b$, $c$ to $b$, etc.

For the math to make sense, the electron must exist somewhere in the universe. Therefore, if you integrate the probability density across all of space (from $-\infty$ to $+\infty$), the total probability must equal exactly $1$ (or $100\%$):

$$\int_{-\infty}^{+\infty} |\psi(x,t)|^2 dx = 1$$

This is called **normalization**. It is used before calculating the probability to solvie for the value of the amplitude constant $A$, also called the normalization constant, of the wave function. This normalization constant adjusts the scale so that the total probability of finding a particle anywhere in space equals exactly $100\%$ not $500\%$ or $50\%$.

Yet, when we normalize $\psi(x,t)$ as derived above, we get $A = 0$, thus $\psi(x) = 0$. The particle doesn't exist mathematically. To solve this, scientists use two special cases:
1. **Particle in a Box** to describe particles that are trapped by forces and boundary conditions.
2. **The Wave Packet** to represent a single, individual particle traveling through empty space with zero potential energy.

This time, we will see how the particle in a box approach works in detail. Yet, first, we need to understand the **Time-Independent Shcordinger Equation**, since it is the main equation we will use in our solution.

### **Derivation of TISE 1**
The first approach to derive the **Time-Independent Schrodinger Equation** utilizes the separation of variables. We write the wave function as a product of two functions (variables). This time, for the sake of clarity, let's denote the wave function as $\Psi(x,t)$:

$$\Psi(x,t) = \psi(x)\phi(t)$$

We substitute this back into the main TDSE:

$$i\hbar\frac{\partial}{\partial t}[\psi(x)\phi(t)] = -\frac{\hbar^2}{2m}\frac{\partial^2}{\partial x^2}[\psi(x)\phi(t)]+ V(x)\psi(x)\phi(t)$$

$$i\hbar\frac{\partial\phi(t)}{\partial t}\psi(x) = -\frac{\hbar^2}{2m}\frac{\partial^2\psi(x)}{\partial x^2}\phi(t)+ V(x)\psi(x)\phi(t)$$

We then arrange this experssion so that all the terms involving $t$ are on the left-hand side and those involving $x$ and on the right-hand side. For this to work, we need to assume that the potential energy doesn't change with time, because if it does, when we try to separate $V$, we won't be able to put it on the left-hand side or the right-hand side. This restricts the TISE to systems where the forces don't change with time. Although this excludes various cases, it matches a massive category of real-world physics problems. Thus, the equation becomes:

$$i\hbar\frac{\partial\phi(t)}{\partial t}\frac{1}{\phi(t)} = [-\frac{\hbar^2}{2m}\frac{\partial^2\psi(x)}{\partial x^2}+ V(x)\psi(x)]\frac{1}{\psi(x)}$$

Since the left-hand side doesn't depend on $x$ and the right-hand side doesn't depend on $t$, their common value doesn't depend on $x$ or $t$, rather, it must be a constant we denote as $k$. To understand this, we put the left-hand side as the function $L(t)$ and the right-hand side as $R(x)$:

$$L(t) = R(x)$$

If we change $t$, $L(t)$ changes. $R(x)$ also wants to change its value. However, $R(x)$ is completely blind to $t$. No matter what we do to $t$, $R(x)$ stays exactly the same. Yet, because the equation states $L(t) = R(x)$, the left side is mathematically foridden from changing **independently**. If $L(t)$ actually changes, the equals sign would break because the right hand side didn't move. The same for the right hand side. Because both sides cannot change at all, both are **constants**. For the sake of depth and mathematical beauty, we can prove this as follows:

$$L(t) = R(x)$$

$$\frac{d}{d x}L(x) = \frac{d}{d x}R(x)$$

Because $L(t)$ has no $x$-dependence, its derivative with respect to $x$ is zero:

$$0 = \frac{d}{dx}R(x)$$

If the derivative of $R(x)$ is zero, by fundamental calculus, that function must be equal to a constant number.

$$\therefore R(x) = k \qquad L(t) = k$$

We separate both sides and set each equal to $k$:

$$i\hbar\frac{1}{\phi(t)}\frac{d\phi(t)}{dt} = k \implies i\hbar\frac{d\phi(t)}{dt} = k\phi(t) \qquad Eq. 1$$

$$\frac{1}{\psi(x)}\left[-\frac{\hbar^2}{2m}\frac{d^2\psi(x)}{dx^2}+ V(x)\psi(x)\right] = k \implies -\frac{\hbar^2}{2m}\frac{d^2\psi(x)}{dx^2}+ V(x)\psi(x) = k\psi(x) \qquad Eq. 2$$

To know what $k$ means, we look closely at $Eq. 2$, the left-hand side represents the operations for kinetic energy plus potential energy actin on the wave function, which gives us the total energy. Therefore, the **Time-Independent Schrodinger Equation** is:

$$\color{#89CFF0}{\bbox[10px, border: 3px solid]{E\psi(x) = -\frac{\hbar^2}{2m}\frac{d^2\psi(x)}{dx^2} +V(x)\psi(x)}}$$

### **Derivation of TISE 2**
The second approach to derive the **Time-Independent Schrodinger Equation** is **The Modern Postulate Approach** that depends mainly on operators.

Before we start, we should understand **operators**. Operators are machines that modify and process **functions**. They are like filters or effects in a photo editing app. We give it the image, adn we get a modified version. There are hats (like $\hat{x})$ on operators to direct your attention to their presence and effects.

Examples of operators include
1. The Derivative Operator ($\frac{d}{dx}$): when we feed it $f(x) = x^3$, we get a brand new function $3x^2$.
2. The Position Operator ($\hat{x}$): it takes a function and multiplies it by $x$. If you give it $\psi$, you get $x\psi$.

We start with a non-relativistic classical particle of mass $m$ moving in a one-dimensional potential energy field $V(x)$. The total mechanical energy $E$ of this system is the sum of its kinetic energy $K$ and its potential energy$V$:

$$E = K + V$$

We then express kinetic energy in terms of momentum, which converts the energy equation into a **Hamiltonian Function**. The hamiltonian system depends on position ($x$) and momentum ($p$):

$$H(x,p) = \frac{p^2}{2m} + V(x)$$

To transition from classical mechanics to quantum mechanics, we change the variables into their corresponding operators.

$$\hat{x} = x$$

$$\hat{p} = -i\hbar \frac{d}{dx}$$

When we substitute the classical variables with the operators, we get a hamiltonian operator itself, since all of its terms are operators.

$$\hat{H} = -\frac{\hbar^2}{2m}\frac{d^2}{dx^2} + V(x)$$

Before we proceed, we need to learn more about **eigenvalues** and **eigenfunctions**.

The prefix **eigen-** is a German word that translates to **own**, **inherent**, or **characteristic**. We know that when an operator processes a wave function, it usually twists and distorts its shape. However, for every operator, there are unique functions that cannot be distorted by that operator. When the operator acts on the function, its shape stays the same. These unique functions are called **eigenfunctions**, the functions which keep their characteristics, or the own characteristic functions. As a result of the operator process, the function gets multiplied by a constant called the **eigenvalue**.

$$\mathrm{Operator \cdot Eigenfunction = Eigenvalue \cdot Eigenfunction}$$

Fortunately, it turns out that $\psi$ is an eigenfunction, and the eigenvalue resulted from multiplying the hamiltonian operator by the wave function is **Energy**;

$$\hat{H}\psi(x) = E\psi(x)$$

To understang why, we should look at **Postulate III (The Measurement Postulate)**. It says that that if a physical observable (a physical quantity that can be empirically detected) is measured, the only possible experimental values that can be obtained are the eigenvalues of that observable's operator. And we know that eigenvlaues are constants; therefore, if we want to find the states of the system where the total energy has a constant value, the wave function must be an eigenfunction of the hamiltonian operator.

Finally, we substitute the hamiltonian differential expression into the eigenvaleu equation:

$$\left[-\frac{\hbar^2}{2m}\frac{d^2}{dx^2} + V(x)\right]\psi(x) = E\psi(x)$$

This gives us the final **Time-Independent Schrodinger Equation**

$$\color{#89CFF0}{\bbox[10px, border: 3px solid]{E\psi(x) = -\frac{\hbar^2}{2m}\frac{d^2\psi(x)}{dx^2} +V(x)\psi(x)}}$$

### **The Particle in a Box Solution**

We imagine a 1D box on the $x$-axis. We set our boundaries to be $x = 0$ and $x = a$, where $0 \le x \le a$.

Inside the boundaries, for $0 \le x \le$, potential energy is zero. Meanwhile, outside the boundaries, potential energy is infinite. Becasue the particle cannot possess an infinite amount of energy to overcome this potential barrier, it has a **zero probability of existence** in these regions. In other words, it doesn't exist **at** or **outside** those boundaries, and $\psi(x) = 0$.

There is a logical quesiton someone may ask: why does $\psi(x) = 0$ at the boundaries?

The wave function is related to the probability of finding the electron at a specific position given the time. We know that outside the boundaires the function must be zero. If we just give the function any other value rather than zero the boundaires, the function will jump from this value to zero outside the boundaires. This makes the function **discontinuous**, contradicting the fact that the probability **must be continuous**. The only solution to prevent the function from causing a catastrophic problem in the universe's probabmility map, the **wave function must be continuous** and smoothly touch zero exactly at the walls. Now, we can continue ...

To find a solution to this specific situation, we set a general wave function $\psi(x)$:

$$\psi(x) = A\sin(kx) + B\cos(kx)$$

At $x = 0$:

$$A\sin(0) + B\cos(0) = 0$$

$$A\sin(0) = 0 \Longrightarrow \sin(0) = 0$$

$$B\cos(0) = 0 \Longrightarrow \cos(0) = 1$$

$$\therefore B = 0$$

This means that the whole term of cosine will be eliminated, and the wave fucntion becomes:

$$\psi(x) = A\sin(kx)$$

At $x = a$:

$$A\sin(ka) = 0$$

If we just set $A = 0$, then $\psi(x) = 0$, and the electron disappears inside the box. Therefore, we conclude that the sine function itself must equal zero:

$$\sin(ka) = 0$$

If we remember the unit circle, the sine function equals zero only when we substitute with multiplications of $\pi$.

$$ka = n\pi \qquad \text{where }n \in \{ 1, 2, 3 ...\}$$

$$k = \frac{n\pi}{a}$$

$$\therefore \psi(x) = A \sin(\frac{n\pi x}{a})$$

This is fascinating, because it tells us that the box boundaries can be only integer multiplications of $\pi$. This also means that the wave inside is a **standing wave**, since only sine functions are capable of creating this type of waves when the boundaires are integer multiplications of $\pi$. This implies that the electron energy is constant and independent of time; let's derive it.

$$\qquad k = \frac{2\pi}{\lambda} \qquad \text{Eq. 1}$$

From de-Broglie realtion:

$$\qquad p = \frac{h}{\lambda} \qquad \text{Eq. 2}$$

From $\text{Eq. 1}$ and $\text{Eq. 2}$:

$$p = \hbar k$$

Becasue $V(x) = 0$:

$$ E = \frac{p^2}{2m} = \frac{\hbar^2k^2}{2m}$$

$$k^2 = \frac{2mE}{\hbar^2}$$

$$\frac{2mE}{\hbar^2} = \frac{n^2\pi^2}{a^2}$$

The final equation for the constant energy of the electron inside the box is:

$$\quad \color{#89CFF0}{\bbox[10px, border: 3px solid]{E_n = \frac{n^2\pi^2\hbar^2}{2ma^2}} } \quad \text{for } n = 1, 2, 3\cdots$$

To find the equation for the probability density function, we normalize the wave function:

$$\int_0^a|\psi(x)|^2 \,dx  =1$$

$$\int_0^a\left[A\sin\left(\frac{n\pi x}{a}\right)\right]^2\, dx = 1$$

We pull out the constant $A^2$ and use the trigonometric identity $\sin(\theta) = \frac{1 - \cos(2\theta)}{2}$:

$$A^2 \int_0^a\frac{1 - \cos\left(\frac{2n\pi x}{a}\right)}{2}\,dx = 1$$

$$\frac{A^2}{2}\left[\int_0^a1\,dx - \int_0^a\cos\left(\frac{2n\pi x}{a}\right)\,dx\right] = 1$$

We use $u$ substitution to solve the second integral:

$$\frac{A^2}{2}\left[x - \frac{a}{2n\pi}\sin\left(\frac{2n\pi x}{a}\right)\right]_0^a = 1$$

At $x = a$, the sine term becomes $\sin(2n\pi) = 0$.

At $x = 0$, everything is $0$.

$$\frac{A^2}{2} [a - 0] = 1$$

$$ \frac{A^2a}{2} = 1$$

$$A = \sqrt{\frac{2}{a}}$$

Therefore, the final probability density function is:

$$\color{#89CFF0}{
  \bbox[10px, border: 3px solid]{
    |\psi(x)|^2 = \frac{2}{a}\sin^2\left(\frac{n\pi x}{a}\right)
  }
}$$






```python
import numpy as np
import matplotlib.pyplot as plt
# Importing the constants pi, the reduced Planck's constant, and the electron mass
from scipy.constants import pi, hbar, m_e

plt.rcParams['figure.dpi'] = 300
```

```python
N = 3
a = 0.53e-10    # 0.5 angstroms

# Plotting energy vs. the quantum number n
# We use the energy equation we derived above
nRange = np.linspace(1, N, N)
nRange2 = np.linspace(0, N, 10000)
E_n = (hbar**2 * pi**2 * nRange**2)/(2 * m_e * a**2)
E_n2 = (hbar**2 * pi**2 * nRange2**2)/(2 * m_e * a**2)
E_n /= 1.602e-19    # We convert the unit to eV
E_n2 /= 1.602e-19

plt.figure()
plt.scatter(nRange, E_n, s = 25, marker = '*')
plt.plot(nRange2, E_n2)
plt.grid(True, alpha = 0.3)
plt.title('Particle in a Box Energy v. Quantum Number')
plt.ylabel('Energy (eV)')
plt.xlabel('Quantum Number (n)')
plt.show()

# Separating between both figures
plt.tight_layout()

# Also, plotting the probability density for each quantum number n vs. the box length x
# Note that the box length is the same as the periemeter of the energy level of the quantum number n
plt.figure()
x = np.linspace(0, a, 1000)
for n in nRange:
  pd = (2/a) * np.sin((n * pi * x)/(a))**2    # Probability density

  plt.plot(x * 1e10, pd, label = f'n = {int(n)}')
  plt.grid(True, alpha = 0.3)
  plt.title('Probability Density Across the Box for each Quantum Number')
  plt.ylabel('Probability density')
  plt.xlabel('x (angstroms)')

plt.legend(loc = 'upper right')
plt.show()

```
